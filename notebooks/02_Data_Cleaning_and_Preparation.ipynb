```python
# Convert date columns in orders dataset to datetime and Load it.
orders = pd.read_csv(r"../data/raw/olist_orders_dataset.csv", parse_dates=['order_purchase_timestamp', 'order_approved_at','order_delivered_customer_date', 'order_estimated_delivery_date','order_delivered_carrier_date'])
```

```python
# Convert date columns in order items dataset to datetime and Load it.
order_items = pd.read_csv(r"../data/raw/olist_order_items_dataset.csv",parse_dates=['shipping_limit_date'])
```

```python
# Convert date columns in reviews dataset to datetime and Load it.
reviews = pd.read_csv(r"../data/raw/olist_order_reviews_dataset.csv",parse_dates=['review_creation_date','review_answer_timestamp'])
```

```python
# Load customers, products,payments,reviews and geolocation datasets.
customers   = pd.read_csv(r"../data/raw/olist_customers_dataset.csv")
products    = pd.read_csv(r"../data/raw/olist_products_dataset.csv")
payments    = pd.read_csv(r"../data/raw/olist_order_payments_dataset.csv")
geolocation = pd.read_csv(r"../data/raw/olist_geolocation_dataset.csv")
sellers     = pd.read_csv(r"../data/raw/olist_sellers_dataset.csv")
```

```python
# Check for null values in "product_category_name" column in products dataset.
products[products["product_category_name"].isnull()]
```

```python
# Replace null values in "product_category_name" with "Unknown" in products dataset.
products.loc[products["product_category_name"].isnull(),"product_category_name"] = "Unknown"
```

```python
# Find null values in "product_weight_g" column in products dataset.
products[products["product_weight_g"].isnull()]
```

```python
# Remove the row with missing information in products dataset.
idx = products[products["product_id"] == "5eb564652db742ff8f28759cd8d2652a"].index
products = products.drop(index= idx)
```

```python
# Check for null values in products dataset after removing the row with missing information.
products.isnull().sum()
```

```python
# Check for null values in "order_approved_at" column in orders dataset.
orders[orders["order_approved_at"].isnull()]
```

```python
# Check the distribution of order status for rows with missing approval dates.
orders[orders["order_approved_at"].isnull()].value_counts(orders["order_status"])
```

```python
# Check for rows with "delivered" order status but missing approval dates.
orders[(orders["order_status"] == "delivered") & (orders["order_approved_at"].isnull())]
```

14 delivered orders have missing approval time, This appears to be a data quality issue rather than a business event.
This records will be retained,but analyses involving approval time should exclude them.

```python
# Check for rows that have missing delivered carrier dates.
orders[orders["order_delivered_carrier_date"].isnull()]
```

```python
# Check the distribution of order status for rows with missing delivered carrier dates.
orders[orders["order_delivered_carrier_date"].isnull()].value_counts(orders["order_status"])
```

```python
# Check for rows with "delivered" order status but missing delivered carrier dates.
orders[(orders["order_delivered_carrier_date"].isnull()) & (orders["order_status"] == "delivered")]
```

Missing values in 'order_delivered_customer_date' are expected

```python
# Create a new column in products table named volume.
products["volume_cm3"] = products["product_length_cm"] * products["product_height_cm"] * products["product_width_cm"] 
```

```python
# Create a new column in orders table named purchase year.
orders["purchase_year"] =  orders["order_purchase_timestamp"].dt.year
```

```python
# Create a new column in orders table named purchase month.
orders["purchase_month"] = orders["order_purchase_timestamp"].dt.month
```

```python
# Create a new column in orders table named purchase month name.
orders["purchase_month_name"] = orders["order_purchase_timestamp"].dt.month_name()
```

```python
# Find orders that have zero payments.
payments[payments["payment_value"] == 0]
```

There are 9 records with a payment value of zero—a very small fraction of the total payment records.
so we will not delete them at this stage without first verifying the nature of these cases.
